```python
import seaborn as sns
```

```python
df=sns.load_dataset('iris')
```

```python
df.head()
```

```python
X=df.drop('species',axis=1)
y=df['species']
```

```python
X,y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(
    X,y,test_size=33,random_state=42)
```

```python
X_train
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
from inspect import Parameter
#Preprunning
Parameter={
'criterion':['log_loss','gini','entropy'],
'splitter':['best','random'],
'max_depth':[None,1,2,3,4,5],
'max_features':['auto','sqrt','log2']
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
treemodel=DecisionTreeClassifier()
cv=GridSearchCV(treemodel,param_grid=Parameter,cv=5,scoring='accuracy')#if you are want to train the data 1000 the cv =5 is helped to the spliting it to the 1000/5 = 200 so that your 200 is your test the data and 800 is your traing the data
cv.fit(X_train,y_train)
```

```python
#Choosing the spicific
cv.best_params_
{
 'ccp_alpha' : 1,
 'criterion' : 'gini',
 'max_depth' : 5,
 'max_features':'auto',
 'splitter':'best'
}
```

```python
cv.predict(X_train)
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
y_pred = cv.predict(X_test)
score = accuracy_score(y_test, y_pred)
score
```

```python
print(classification_report(y_pred,y_test))
```
