```python
df=df[df['species'] != 'setosa']
```

```python
df['species']=df['species'].map({'versicolor':0,'virginica':1})
```

```python
X=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
X
```

```python
#Traing and Tezting data
from sklearn.model_selection import  train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
classifier=LogisticRegression()
```

```python
from inspect import Parameter
from sklearn.model_selection import GridSearchCV
Parameter={'penalty':['l1','l2','elasticnet'],'c':[1,2,3,4,5,6,10,20,30,40,50],'max_tier':[100,200,300]}
```

```python
classifier_regression=GridSearchCV(classifier,param_grid=Parameter,scoring='accuracy',cv=5)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV

# Corrected parameter dictionary: 'C' instead of 'c', 'max_iter' instead of 'max_tier'
# Also, specifying 'solver': ['saga'] to ensure compatibility with 'l1' and 'elasticnet' penalties.
corrected_Parameter = {
    'penalty': ['l1','l2','elasticnet'],
    'C': [1,2,3,4,5,6,10,20,30,40,50],
    'max_iter': [100,200,300],
    'solver': ['saga'], # 'saga' solver supports l1, l2, and elasticnet penalties
    'l1_ratio': [0.5] # Required for 'elasticnet' penalty when using 'saga' solver
}

# Re-instantiate the classifier (ensuring it's the base LogisticRegression)
classifier = LogisticRegression(random_state=42) # Added random_state for reproducibility

# Re-create GridSearchCV with the corrected parameter grid
classifier_regression = GridSearchCV(classifier, param_grid=corrected_Parameter, scoring='accuracy', cv=5)

# Fit the model with the corrected parameters
classifier_regression.fit(X_train,y_train)
```

```python
classifier_regression.fit(X_train,y_train)
```

```python
print(classifier_regression.best_params_)
```

```python
print(classifier_regression.best_score_)
```

```python
y_pred=classifier_regression.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
score=accuracy_score(y_pred,y_test)
print(score)
```
